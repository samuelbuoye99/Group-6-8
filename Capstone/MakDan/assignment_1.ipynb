## A student writes this function to add the integers from 1 through n:

```python
def total(n):
    if n == 1:
        return 1
    return n + total(n - 1)
```

### B. Find the problem: What happens for total(0) and total(-3)? Rewrite the function so negative inputs raise ValueError and zero returns 0. State the behavior you chose for non-integer input.

```python
def total(n):
    if isinstance(n, bool) or not isinstance(n, int):
        raise TypeError("n must be an integer")

    if n < 0:
        raise ValueError("n must not be negative")

    if n == 0:
        return 0

    if n == 1:
        return 1

    return n + total(n - 1)
```

### D. Code it: Write countdown_sum(n), which recursively returns the sum of odd positive integers from n down to 1. Return 0 for n <= 0. Do not use loops, sum(), count(), or comprehensions.
Examples:

```python
def countdown_sum(n):
    if isinstance(n, bool) or not isinstance(n, int):
        raise TypeError("n must be an integer")

    if n <= 0:
        return 0
    if n % 2 == 0:
        return countdown_sum(n-1)
    return n + countdown_sum(n-2)
```

```python
print(countdown_sum(5))
print(countdown_sum(4))
print(countdown_sum(1))
print(countdown_sum(0))
```

# QUEST 2 - Big O: Growth, Not Guesswork
Learning target: Connect Python operations and loops to time complexity, auxiliary space, and practical trade-offs.

```python
# Implementation A
def has_duplicate_a(items):
    for i in range(len(items)):
        for j in range(i + 1, len(items)):
            if items[i] == items[j]:
                return True
    return False

# Implementation B
def has_duplicate_b(items):
    seen = set()
    for item in items:
        if item in seen:
            return True
        seen.add(item)
    return False

# Implementation C
def has_duplicate_c(items):
    items = sorted(items)
    for i in range(1, len(items)):
        if items[i] == items[i - 1]:
            return True
    return False
```

# QUEST 3 — Hashing and Hash Maps: Build the Mechanism
Learning target: Understand keys, hash indexes, collisions, buckets, and why a hash map still needs to compare keys.

```python
def hash_key(key):
    return sum(ord(char) for char in key) % 7
```

### C. Implement: Build a seven-slot hash map using separate chaining. Implement put(key, value), get(key), and delete(key). put must update an existing key rather than duplicate it; get must return None for a missing key; delete must remove only the requested key.

```python
def hash_key(key):
    if not isinstance(key, str):
        raise TypeError("Keys must be strings")

    return sum(ord(char) for char in key) % 7


class HashMap:
    def __init__(self):
        self.buckets = [[] for _ in range(7)]

    def put(self, key, value):
        index = hash_key(key)
        bucket = self.buckets[index]

        for i in range(len(bucket)):
            if bucket[i][0] == key:
                bucket[i] = (key, value)
                return

        bucket.append((key, value))

    def get(self, key):
        index = hash_key(key)
        bucket = self.buckets[index]

        for stored_key, value in bucket:
            if stored_key == key:
                return value

        return None

    def delete(self, key):
        index = hash_key(key)
        bucket = self.buckets[index]

        for i in range(len(bucket)):
            if bucket[i][0] == key:
                bucket.pop(i)
                return True

        return False
```

```python
table = HashMap()

table.put("cat", 10)
table.put("act", 20)
table.put("dog", 30)
table.put("god", 40)

print(table.get("cat"))
print(table.get("act"))
print(table.get("dog"))
print(table.get("god"))
print(table.get("bird"))
```

# QUEST 4 — Decorators: Wrap Behavior Without Breaking It
Learning target: Understand function objects, decoration time, wrapper calls, arguments, and return values

```python
def announce(func):
    print("Decorator is running")

    def wrapper(*args, **kwargs):
        print("Before function")
        result = func(*args, **kwargs)
        print("After function")
        return result

    return wrapper

@announce
def multiply(a, b):
    print("Multiplying")
    return a * b

print("Program started")
answer = multiply(3, 4)
print("Answer:", answer)
```
